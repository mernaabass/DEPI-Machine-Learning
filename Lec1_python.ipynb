# Questions on Python Data Types

1. Create a list containing the numbers 1 to 5. Then:
   - Append the number 6 to the list.
   - Replace the second element with the number 10.
   - Print the list after the changes.

2. Given the list `my_list = ['apple', 'banana', 'cherry', 'date']`, perform the following:
   - Access the first element in the list.
   - Remove the last element from the list.
   - Insert the string `'grape'` at index 2.
   - Print the modified list.

3. Create a set containing the numbers 1, 2, 3, 4. Then:
   - Add the number 5 to the set.
   - Try to add the number 3 to the set again. What happens?

4. Given the set `my_set = {'red', 'green', 'blue'}`, perform the following:
   - Check if `'green'` is present in the set.

   - Remove `'red'` from the set.
   - Print the modified set.

5. Create a tuple containing the numbers 1, 2, 3, 4, 5. Then:
   - Access the third element of the tuple.
   - Try to modify the second element of the tuple. What happens?

6. Given the tuple `my_tuple = ('apple', 'banana', 'cherry')`, perform the following:
   - Access the last element in the tuple.

7. Create a dictionary with the following key-value pairs: `'name': 'John', 'age': 25, 'city': 'New York'`. Then:
   - Access the value associated with the key `'name'`.
   - Change the value of the key `'age'` to 26.
   - Add a new key `'job'` with the value `'Engineer'`.
   - Print the modified dictionary.

8. Given the dictionary `my_dict = {'brand': 'Toyota', 'model': 'Corolla', 'year': 2019}`, perform the following:
   - Access the value associated with the key `'model'`.
   - Remove the key `'year'` from the dictionary.
   - Add a new key `'color'` with the value `'blue'`.
   - Print the modified dictionary.

```python
# Create a list containing numbers 1 to 5
numbers = [1, 2, 3, 4, 5]

# Append the number 6 to the list
numbers.append(6)

# Replace the second element (index 1) with 10
numbers[1] = 10

# Print the list
print(numbers)
```

```python
my_list = ['apple', 'banana', 'cherry', 'date']

# Access the first element
first_element = my_list[0]
print("First element:", first_element)

# Remove the last element
my_list.pop()  # or del my_list[-1]

# Insert 'grape' at index 2
my_list.insert(2, 'grape')

# Print the modified list
print("Modified list:", my_list)
```

```python
# Create a set containing 1, 2, 3, 4
num_set = {1, 2, 3, 4}

# Add the number 5
num_set.add(5)

# Try to add the number 3 again
num_set.add(3)

# Remove the number 2
num_set.remove(2)

# Print the final set
print(num_set)
```

```python
my_set = {'red', 'green', 'blue'}

# Check if 'green' is present
is_green_present = 'green' in my_set
print("Is 'green' in set?:", is_green_present)

# Add 'yellow'
my_set.add('yellow')

# Remove 'red'
my_set.remove('red')

# Print the modified set
print("Modified set:", my_set)
```

```python
# Create a tuple containing 1, 2, 3, 4, 5
my_tuple = (1, 2, 3, 4, 5)

# Access the third element (index 2)
third_element = my_tuple[2]
print("Third element:", third_element)

# Try to modify the second element
try:
    my_tuple[1] = 10
except TypeError as e:
    print("Error when modifying tuple:", e)

# Print the tuple
print("Tuple:", my_tuple)
```

```python
my_tuple = ('apple', 'banana', 'cherry')

# Access the last element
last_element = my_tuple[-1]
print("Last element:", last_element)

# Check if 'banana' is in the tuple
is_banana_present = 'banana' in my_tuple
print("Is 'banana' in tuple?:", is_banana_present)

# Print the tuple
print("Tuple:", my_tuple)
```

```python
# Create a dictionary
person = {'name': 'John', 'age': 25, 'city': 'New York'}

# Access value of key 'name'
name_value = person['name']
print("Name:", name_value)

# Change value of key 'age' to 26
person['age'] = 26

# Add a new key 'job' with value 'Engineer'
person['job'] = 'Engineer'

# Print modified dictionary
print("Modified dictionary:", person)
```

```python
my_dict = {'brand': 'Toyota', 'model': 'Corolla', 'year': 2019}

# Access value associated with key 'model'
model_value = my_dict['model']
print("Model:", model_value)

# Remove the key 'year'
del my_dict['year']  # or my_dict.pop('year')

# Add new key 'color' with value 'blue'
my_dict['color'] = 'blue'

# Print modified dictionary
print("Modified dictionary:", my_dict)
```

## 💻 Exercises ➞ <span class='label label-default'>Dictionary</span>

1. Create an empty dictionary called **`bird`**
2. Add **`name`**, **`color`**, **`breed`**, **`legs`**, **`age`** to the **`bird`** dictionary
3. Create a **`student`** dictionary and add **`first_name`**, **`last_name`**, **`gender`**, **`age`**, **`marital_status`**, **`skills`**, **`country`**, **`city`** and **`address`** as keys for the dictionary
3. Get the length of the **`student`** dictionary
4. Get the value of **`skills`** and check the data type, it should be a list
5. Modify the **`skills`** values by adding one or two skills
6. Get the dictionary keys as a list
7. Get the dictionary values as a list
8. Change the dictionary to a list of tuples using **`items()`** method
9. Delete one of the items in the dictionary
10. Delete one of the dictionaries


```python
# 1. Create an empty dictionary called bird
bird = {}

# 2. Add name, color, breed, legs, age to the bird dictionary
bird['name'] = 'Eagle'
bird['color'] = 'Brown'
bird['breed'] = 'Golden Eagle'
bird['legs'] = 2
bird['age'] = 4

# 3. Create a student dictionary and add keys
student = {
    'first_name': 'Merna',
    'last_name': 'Abass',
    'gender': 'Female',
    'age': 22,
    'marital_status': 'Single',
    'skills': ['Python', 'Dart', 'Flutter', 'Machine Learning'],
    'country': 'Egypt',
    'city': 'Damietta',
    'address': 'Main Street'
}

# 4. Get the length of the student dictionary
student_length = len(student)
print("Student Dictionary Length:", student_length)

# 5. Get the value of skills and check the data type
skills_value = student['skills']
print("Skills Value:", skills_value)
print("Skills Data Type:", type(skills_value))  # Should be <class 'list'>

# 6. Modify the skills values by adding one or two skills
student['skills'].append('TensorFlow')
student['skills'].append('FastAPI')
print("Updated Skills:", student['skills'])

# 7. Get the dictionary keys as a list
student_keys = list(student.keys())
print("Keys List:", student_keys)

# 8. Get the dictionary values as a list
student_values = list(student.values())
print("Values List:", student_values)

# 9. Change the dictionary to a list of tuples using items() method
student_tuples = list(student.items())
print("List of Tuples:", student_tuples)

# 10. Delete one of the items in the dictionary
del student['address']  # or student.pop('address')
print("After Deleting Address Item:", student)

# 11. Delete one of the dictionaries
del bird
# print(bird) # Executing this now will raise a NameError because 'bird' is completely deleted.
```
